O algoritmo DC é uma forma de como você pensa sobre o problema. Nós precisamos descobrir o caso-base do problema, que deve ser o mais simples possível. Para isso, precisamos 



```python
def soma(lista):
    total = 0
    for x in lista:
        total += x
    return total
print(soma([1, 2, 3, 4]))
```

```python
def itens_total(lista:list): 
    if len(lista) == 0:
        return 0
    
    return 1 + itens_total(lista[1:])
    
        
lista = [1, 2, 3, 4, 5]
a = itens_total(lista)
print(a)
```
