*** Quicksort  ***

é um algoritmo de ordenação extremamente rápido! Ele usa a estratégia DC.



Basicamente, nós selecionamos um elemento como o pivô, neste caso, sendo o caso-base para o DC. Além disso, ao pegar um elemento como referência, podemos concluir quais são os outros que são maiores ou menores em relação ao nosso modelo. Cada classificação vai ser um array, ou seja, separamos o array principal em dois (em relação ao pivô). Chamamos isso de partição. Entretanto, os subarrays ainda não estão ordenados!


Para ordernar os subarrays, basta aplicar o quicksort novamente! Isso serve para array de qualquer tamanho! 



```python
def quicksort(array):
    if len(array) < 2:
        return array 
    else:
        pivo = array[0] 
        menores = [i for i in array[1:] if i <= pivo] 
        maiores = [i for i in array[1:] if i > pivo] 
        return quicksort(menores) + [pivo] + quicksort(maiores)
```
