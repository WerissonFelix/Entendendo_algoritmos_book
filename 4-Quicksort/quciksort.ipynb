*** Quicksort  ***

é um algoritmo de ordenação extremamente rápido! Ele usa a estratégia DC.



Basicamente, nós selecionamos um elemento como o pivô, neste caso, sendo o caso-base para o DC. Além disso, ao pegar um elemento como referência, podemos concluir quais são os outros que são maiores ou menores em relação ao nosso modelo. Cada classificação vai ser um array, ou seja, separamos o array principal em dois (em relação ao pivô)
