***Funções Hash***



Funções Hash convertem uma string em um número específico. Essa conversão não é aleatória, ou seja, se dois inputs tiverem a mesma string, os outputs serão os mesmos.

Isso permite que buscas e inserções sejam extremamente eficientes!!!

Um ótimo exemplo: Agenda telefônica. Para cada contato, há um nome e um telefone. Caso já exista um contato o nome dado no input, o output será o telefeone desse contato. Contudo, se esse contato não existir, será criado, relacionando um nome com um número.

Em python, tables hash são dicionários.

```python
# Exemplo da Agenda telefônica


lista_telefonica = dict()  # pode ser lista_telefonica = {} tbm

lista_telefonica["jenny"] = 8675309
lista_telefonica["emergency"] = 911

print(lista_telefonica["jenny"])
```

*Evitar entradas duplicadas*

Podemos evitar duplicas com o dict.get(name). Ele retorna um value caso exista a relação entre o input e algum outro value, caso o contrário, retorna None. 

Isso é bom para evitar duplicadas e fazer verificações

```python
# Exemplo de uma função que verifica se uma pessoa já voltou ou não

voted = {}
def verifica_eleitor(nome):
    """  
        Retorna 'Mande embora!' se a pessoa já votou;
        Retorna 'Pode votar!' se a pessoa não votou ainda;
    """
    if voted.get(nome):
        print ("Mande embora!")
    else:
        voted[nome] = True
        print ("Deixe votar!")

verifica_eleitor("tom")

verifica_eleitor("mike")

verifica_eleitor("mike")

# A gente tenta colocar mike dnv, mas ele é barrado!!
```
