***Colisões***


Acontecem quando a função Hash retorna o mesmo output para diferentes inputs, nos obrigando a recorrer a algumas alternativas.

Como usar listas encadeadas nas posições, por exemplo.


A ideia é justamente evitar isso, porque listas encadeadas muito longas podem levar a tabela hash ao seu pior caso o O(n). 

Podemos evitar fazendo uma boa função hash, distribuindo as chaves uniformemente pela tabela.
