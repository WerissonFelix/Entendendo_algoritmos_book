*** RECURSÃO ***

Recursão é quando uma função chama a si mesma.


**Caso-base e caso recursivo**

Caso recursivo: quando a função chama a si mesma

Caso-base: quando a função não chama a si mesma novamente, evitando um loop infinito.




```python
# Caso-base e caso recursivo
def regressiva(i):
    print i
    if i <= 1: 
        return
    else: 
        regressiva(i-1)
```

*** A PILHA ***

 Uma estrutura de dados simples, a qual contém apenas duas ações: push e pop. Push para adicionar um novo item no topo da lista, e pop para remover e ler o item do topo.


**PILHA DE CHAMADA** 

Quando uma função chama outras funções. Durante este processo, as variáveis das funções são guardadas em uma pilha, em ordem de chegada, além disso, quando uma função retorna (ou seja, acaba) suas variáveis são retiradas dessa pilha na memória.
