***NOTAÇÃO BIG O***


É uma naotação que diz o quão rápido é um algoritmo. 


Algoritmos de pesquisa simples e binárias possuem taxas de crescimento de velocidade de execução diferentes!

É necessário saber se o tempo de execução aumenta conforme a lista aumenta!


Uma notação Big O não informa quantos segundos foram necessários para a execução completa, mas sim o número de operações!




Exemplos de tempo de execução Big O:

- O(log n): Conhecido como tempo logarítmico, o mais rápido de todos

- O(n): conhecido como tempo linear e também como o pior cenário possível para uma pesquisa simples.

- O(n * log n ): Algoritmos de organização, ordenação quicksort
