**Pesquisa binária**

A pesquisa binária é um algoritmo (Conjunto de instruções que realizam uma tarefa) que faz uma busca em uma lista ordenada, para encontrar determinados elementos.

Resume-se em uma busca inteligente por eliminatória. Ao invés de procurar cegamente elemento por elemento dentro da lista, optiamos por posições (index) estratégicas deles, buscando reduzir a quantidade de possibilidades/opções.




Essas imagens representam perfeitamente a lógica da pesquisa binária. Reduzir o número de tentativas é crucial para otimizações em sistemas de busca! O que, obviamente faz toda diferença em vários sentidos!
