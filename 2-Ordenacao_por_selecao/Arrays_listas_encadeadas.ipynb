***Arrays***


Quando falamos de memória, todos os itens de um array são alocados um ao lado do outro. Caso um novo elemento seja adicionado E não exista um espaço da "casa" ao lado, então todos elementos são alocados para um novo lugar na memória, pois todos elementos de um array precisam ser "vizinhos" nas "gavetas" da memória. Ou seja, este processo pode ser lento, uma vez que sempre pode haver esta deslocação de todos os itens.

Uma possível solução para isto é fazer uma reserva na memória, para prever futuros novos itens. Entretanto, esses novos itens podem não aparecer, e então, espaço na memória estariam sendo desperdiçados!



Contudo, uma vantagem é que arrays são muito bons para selecionar itens aleatórios ou então, o último item do grupo. Como todos os itens são vizinhos uns dos outros, já sabemos exatamente qual é o endereço de cada elemento do array, diferentemente das listas encadeadas.




Para deletar elementos ou então adicionar, listas encadeadas são muito melhores, pois basta alterar o endereço de referência do elemento anterior

***Listas Encadeadas***

Nas listas encadeadas, cada elemento possui o endereço do próximo. Assim, não é necessário que todos os elementos sejam vizinhos uns dos outros, basta que tenha espaço na memória, juntos ou não. Dessa forma, evitando o problema comum dos arrays, movimentação de dados na memória.




Porém, listas encadeadas são terríveis para encontrar itens aleatórios ou então, o último item da lista. Como o endereço de cada item é armazenado no item anterior, seria necessário percorrer todos os itens anteriores ao desejado para descobrir seu endereço.

Podemos juntar esses dois em uma única estrutura: estrutura híbrida com arrays e listas encadeadas. Basicamente, um Array cujo elementos são listas encadeadas
