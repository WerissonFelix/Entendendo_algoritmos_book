***Arrays***


Quando falamos de memória, todos os items de um array são alocados um ao lado do outro. Caso um novo elemento seja adicionado E não exista um espaço da "casa" ao lado, então todos elementos são alocados para um novo lugar na memória, pois todos elementos de um array precisam ser "vizinhos" nas "gavetas" da memória. Ou seja, este processo pode ser lento, uma vez que sempre pode haver esta deslocação de todos os items.

Uma possível solução para isto é fazer uma reserva na memória, para prever futuros novos items. Entretanto, esses novos items podem não aparecer, e então, espaço na memória estariam sendo desperdiçados!
