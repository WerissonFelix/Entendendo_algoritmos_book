***Algoritmos de ordenação***

Logo de cara, em uma lista de números, para orderná-la é necessário percorrer todos elementos da lista, para descobrir qual é o maior elemento. Dito isso, temos no mínimo O(n) em Big O, mas precisamos analisar novamente a lista, para saber qual é o segundo elemento mais rápido, o terceiro e por aí vai.



```python
def buscaMenor(arr):
    menor = arr[0] 
    menor_indice = 0 
    for i in range(1, len(arr)):
        if arr[i] < menor:
            menor = arr[i]
            menor_indice = i
    return menor_indice
```

```python
def ordenacaoporSelecao(arr): 
    novoArr = []
    for i in range(len(arr)):
        menor = buscaMenor(arr) 
        novoArr.append(arr.pop(menor))
    return novoArr
```

```python
print (ordenacaoporSelecao([5, 3, 6, 2, 10]))
```
