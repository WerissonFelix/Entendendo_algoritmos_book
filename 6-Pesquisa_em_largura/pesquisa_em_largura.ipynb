***Filas***

É uma estrutura de dados FIFO (First In, First Out) que funciona exatamente como uma fila na vida real, como uma fila de ônibus, por exemplo.

Não é possível acessar elementos aleatoriamente, somente em ordem de chegada (do primeiro até o último).

Além disso, somente duas operações são possíveis: enqueue (enleirar) e dequeue
(desenleirar).



o Enqueue funciona como um append e o Dequeue funciona como um pop(0)


Perceba que filas são estruturas de dados FIFO, equanto pilhas são LIFO (Last In, First Out)




***IMPLEMENTANDO GRAFOS***

Para implementar um grafo em python, utilizaremos a querida tabela Hash (dict) para representar os vértices e seus respectivos vizinhos.

Vamos representar esse grafo com um código python:


```python
grafo = {}
grafo["voce"] = ["alice", "bob", "claire"]
grafo["bob"] = ["anuj", "peggy"]
grafo["alice"] = ["peggy"]
grafo["claire"] = ["thom", "jonny"]
grafo["anuj"] = []
grafo["peggy"] = []
grafo["thom"] = []
grafo["jonny"] = []
```

```python
from collections import deque

def pessoa_e_vendedor(nome):
    return nome[-1] == 'm'

def pesquisa(nome):
    fila_de_pesquisa = deque()
    fila_de_pesquisa += grafo[nome]
    verificadas = [] 
    while fila_de_pesquisa:
        pessoa = fila_de_pesquisa.popleft()
        if not pessoa in verificadas: 
            if pessoa_e_vendedor(pessoa):
                print (pessoa + " é um vendedor de manga!")
                return True
        else:
            fila_de_pesquisa += grafo[pessoa]
            verificadas.append(pessoa) 
    return False

pesquisa('voce')
```
